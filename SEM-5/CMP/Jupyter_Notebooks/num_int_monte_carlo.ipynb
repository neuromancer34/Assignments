## Numerical Integration using Monte Carlo

### ALGORITHM
1. Set the sample size $N$
2. Generate two independent arrays of N uniform random numbers, r1 and r2 in [0, 1], representing 2D coordinate pairs (x, y) inside a $1 \times 1$ bound box.
3. Evaluate the function at each x: $f(r_1) = \exp(-10(r_1 - 0.5)^2)$.
4. Check which points land under the curve ($r_2 \le f(r_1)$).
5. Count the total accepted points (n) and compute the integral estimate as: 
$$\text{Area} \approx \frac{n}{N} \times \text{Bound Box Area} = \frac{n}{N}$$
6. Filter and store the valid x-coordinates.
7. Plot a normalized histogram of valid x to get the distribution.
8. Compute a continuous array of x from 0 to 1 and evaluate its y.
9. Compute the total numerical area A using trapezoidal integration.
10. Plot the normalized probability density curve $\frac{y(x)}{A}$ over the histogram to visually verify that the random sample distribution matches the target function.
11. Iterate over an increasing sequence of sample sizes n from 10 to N.
12. For each sample size, generate random points and estimate the integral.
13. Plot the estimated integral versus the number of points to demonstrate convergence toward the actual value.

### CODE AND OUTPUT

```python
import matplotlib.pyplot as plt
%matplotlib inline
import numpy as np

N = 10000  # Number of samples

# Generate the random points
r1 = np.random.uniform(0, 1, N)
r2 = np.random.uniform(0, 1, N)

f_x = np.exp(-10 * (r1 - 0.5) ** 2)  # Function

n = np.sum(r2 <= f_x)  # Number of valid points
prob = n / N  # Probabilty/Integral(for Area=1)
valid_x = r1[r2 <= f_x]

plt.hist(valid_x, bins=50, density=True, edgecolor="black")

x_smooth = np.linspace(0, 1, 100)
y_smooth = np.exp(-10 * (x_smooth - 0.5) ** 2)
A = np.trapezoid(y_smooth, x_smooth)  # Area under curve/Integral
n_prob = y_smooth / A  # Scaling Factor

plt.plot(x_smooth, n_prob, "r-", linewidth=2)
plt.show()
print(f"Integral by Monte Carlo: {prob:.3f}")
print(f"Integral by Trapezoidal Method: {A:.3f}")


sample_sizes = range(10, N + 1, 30)
int_estimates = []

for n_pts in sample_sizes:
    r1_c = np.random.uniform(0, 1, n_pts)
    r2_c = np.random.uniform(0, 1, n_pts)
    int_estimates.append(np.mean(r2_c <= np.exp(-10 * (r1_c - 0.5) ** 2)))  # Integral

plt.plot(sample_sizes, int_estimates)
plt.xlabel("Number of samples (N)")
plt.ylabel("Estimated integral")
plt.show()
print(f"Final Integral: {int_estimates[-1]:.3f}")
```
